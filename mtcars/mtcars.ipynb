```python
import pandas as pd
```

```python
df = pd.read_csv('/content/mtcars.csv')
display(df.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Visualize the distribution of 'mpg'
plt.figure(figsize=(8, 6))
sns.histplot(df['mpg'], kde=True)
plt.title('Distribution of MPG')
plt.xlabel('Miles Per Gallon (MPG)')
plt.ylabel('Frequency')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Visualize the relationship between 'hp' and 'mpg'
plt.figure(figsize=(10, 7))
sns.scatterplot(x='hp', y='mpg', data=df, hue='cyl', size='wt', sizes=(20, 400), palette='viridis', alpha=0.7)
plt.title('MPG vs. Horsepower (colored by cylinders, sized by weight)')
plt.xlabel('Horsepower (hp)')
plt.ylabel('Miles Per Gallon (mpg)')
plt.legend(title='Cylinders / Weight')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
