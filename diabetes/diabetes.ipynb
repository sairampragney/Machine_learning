```python
import pandas as pd
```

```python
df = pd.read_csv('/content/diabetes (1).csv')
display(df.head())
```

## Data Visualization

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Visualize the distribution of the 'Outcome' variable
plt.figure(figsize=(6, 4))
sns.countplot(x='Outcome', data=df)
plt.title('Distribution of Diabetes Outcome')
plt.xlabel('Outcome (0: Non-Diabetic, 1: Diabetic)')
plt.ylabel('Count')
plt.show()
```

Next, let's look at the distribution of some important numerical features (Glucose, BMI, Age) for both diabetic and non-diabetic outcomes.

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

sns.histplot(data=df, x='Glucose', hue='Outcome', kde=True, ax=axes[0])
axes[0].set_title('Glucose Distribution by Outcome')

sns.histplot(data=df, x='BMI', hue='Outcome', kde=True, ax=axes[1])
axes[1].set_title('BMI Distribution by Outcome')

sns.histplot(data=df, x='Age', hue='Outcome', kde=True, ax=axes[2])
axes[2].set_title('Age Distribution by Outcome')

plt.tight_layout()
plt.show()
```
