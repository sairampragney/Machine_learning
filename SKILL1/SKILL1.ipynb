```python
import pandas as pd
```

```python
df = pd.read_csv('/content/titanic.csv')
display(df.head())
```

```python
import matplotlib.pyplot as plt

survival_counts = df['survived'].value_counts()

plt.figure(figsize=(7, 7))
plt.pie(survival_counts, labels=['Not Survived (0)', 'Survived (1)'], autopct='%1.1f%%', startangle=90, colors=['salmon', 'lightgreen'])
plt.title('Proportion of Survivors on Titanic')
plt.axis('equal') # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()
```
